```python
X = df.drop("cardio", axis=1)

y = df["cardio"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

```python
svm_model = SVC(
    kernel="rbf",
    random_state=42
)

svm_model.fit(
    X_train_scaled,
    y_train
)

svm_pred = svm_model.predict(X_test_scaled)

svm_accuracy = accuracy_score(
    y_test,
    svm_pred
)

print("SVM Accuracy:", svm_accuracy)
```

```python
knn_model = KNeighborsClassifier(
    n_neighbors=5
)

knn_model.fit(
    X_train_scaled,
    y_train
)

knn_pred = knn_model.predict(X_test_scaled)

knn_accuracy = accuracy_score(
    y_test,
    knn_pred
)

print("KNN Accuracy:", knn_accuracy)
```

```python
dt_model = DecisionTreeClassifier(
    random_state=42
)

dt_model.fit(
    X_train,
    y_train
)

dt_pred = dt_model.predict(X_test)

dt_accuracy = accuracy_score(
    y_test,
    dt_pred
)

print("Decision Tree Accuracy:", dt_accuracy)
```

```python
lr_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

lr_model.fit(
    X_train_scaled,
    y_train
)

lr_pred = lr_model.predict(X_test_scaled)

lr_accuracy = accuracy_score(
    y_test,
    lr_pred
)

print("Logistic Regression Accuracy:", lr_accuracy)
```

```python
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(
    X_train,
    y_train
)

rf_pred = rf_model.predict(X_test)

rf_accuracy = accuracy_score(
    y_test,
    rf_pred
)

print("Random Forest Accuracy:", rf_accuracy)
```

```python
results = pd.DataFrame({
    "Model": [
        "SVM",
        "KNN",
        "Decision Tree",
        "Logistic Regression",
        "Random Forest"
    ],

    "Accuracy": [
        svm_accuracy,
        knn_accuracy,
        dt_accuracy,
        lr_accuracy,
        rf_accuracy
    ]
})

results["Accuracy Percentage"] = results["Accuracy"] * 100

results
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    x="Model",
    y="Accuracy Percentage",
    data=results
)

plt.title("Machine Learning Model Accuracy Comparison")
plt.xlabel("Machine Learning Model")
plt.ylabel("Accuracy (%)")

plt.xticks(rotation=20)

plt.show()
```

```python
import gradio as gr
import pandas as pd

def predict_from_ui(
    gender,
    age_years,
    height,
    weight,
    ap_hi,
    ap_lo,
    cholesterol,
    gluc,
    smoke,
    alco,
    active
):

    input_data = pd.DataFrame([[
        gender,
        height,
        weight,
        ap_hi,
        ap_lo,
        cholesterol,
        gluc,
        smoke,
        alco,
        active,
        age_years
    ]], columns=[
        "gender",
        "height",
        "weight",
        "ap_hi",
        "ap_lo",
        "cholesterol",
        "gluc",
        "smoke",
        "alco",
        "active",
        "age_years"
    ])

    input_scaled = scaler.transform(input_data)

    prediction = svm_model.predict(input_scaled)

    if prediction[0] == 1:
        return "Cardiovascular Disease Present"
    else:
        return "Cardiovascular Disease Absent"


demo = gr.Interface(
    fn=predict_from_ui,

    inputs=[
        gr.Number(label="Gender", value=2),
        gr.Number(label="Age (Years)", value=45),
        gr.Number(label="Height (cm)", value=165),
        gr.Number(label="Weight (kg)", value=70),
        gr.Number(label="Systolic Blood Pressure", value=120),
        gr.Number(label="Diastolic Blood Pressure", value=80),
        gr.Number(label="Cholesterol", value=1),
        gr.Number(label="Glucose", value=1),
        gr.Number(label="Smoking (0/1)", value=0),
        gr.Number(label="Alcohol Intake (0/1)", value=0),
        gr.Number(label="Physical Activity (0/1)", value=1)
    ],

    outputs=gr.Textbox(label="Prediction Result"),

    title="Cardiovascular Disease Prediction System",

    description="Machine Learning based prediction using the final SVM model."
)

demo.launch(share=True)
```
